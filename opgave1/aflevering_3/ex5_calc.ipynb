```python
import numpy as np
import matplotlib.pylab as plt

#f(x,y) = xy + 14 = 1
L1_lim = 1
L1_upper = 14
x = 0.000
y = 0.000

def f(x,y):
    return x*y + 14 

def g(x,y):
    return x**2 + y**2 - 18

x = np.linspace(-15,15,500)
y = np.linspace(-15,15,500)

x_p = [-13, -6.5,-6.5/2,-1, 1, 6.5/2 ,6.5, 13]
y_p = [1, 2, 4, 13, -13, -4,-2, -1]

X, Y = np.meshgrid(x,y)
Z=f(X,Y)
Z2=g(X,Y)
print(Z)
plt.contour(X,Y,Z, levels=[1],colors="blue")
plt.contour(X,Y,Z2, levels=[0],colors="red")
plt.scatter(x_p, y_p, color='black', s=50, zorder=5)
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```
